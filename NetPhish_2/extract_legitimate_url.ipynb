```python
from warcio.archiveiterator import ArchiveIterator
import gzip
import csv
import requests
```

```python
url = "https://data.commoncrawl.org/crawl-data/CC-MAIN-2025-13/segments/1742004432822.9/warc/CC-MAIN-20250315031626-20250315061626-00001.warc.gz"
r = requests.get(url)

with open("CC-MAIN-20250315031626-20250315061626-00000.warc.gz", "wb") as f:
    f.write(r.content)
```

```python
warc_file = 'CC-MAIN-20250315031626-20250315061626-00001.warc.gz'
output_file = 'legitimateB.csv'
max_urls = 50000
count = 0

with open(warc_file, 'rb') as stream, open(output_file, 'w', newline='', encoding='utf-8') as outfile:
    writer = csv.writer(outfile)
    writer.writerow(['URL'])

    try:
        for record in ArchiveIterator(stream):
            if record.rec_type == 'response':
                url = record.rec_headers.get_header('WARC-Target-URI')
                status = record.http_headers.get_statuscode()
                content_type = record.http_headers.get_header('Content-Type')

                if url and status == '200' and content_type and 'text/html' in content_type:
                    writer.writerow([url])
                    count += 1


    except Exception as e:
        print(f"⚠️ Skipping record due to error: {e}")

print(f"✅ Finished. Extracted {count} URLs.")
```
