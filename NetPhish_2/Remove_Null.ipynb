```python
import pandas as pd
import csv

# ===== INPUT / OUTPUT FILES =====
INPUT_CSV  = "/home/tejas/Desktop/phish_dataset/New_NetPhish_Versions/NetPhish_2_V0.csv"
OUTPUT_CSV = "/home/tejas/Desktop/phish_dataset/New_NetPhish_Versions/NetPhish_2_V1.csv"

# ===== STEP 1: READ CSV SAFELY =====
df = pd.read_csv(INPUT_CSV, quoting=csv.QUOTE_ALL, low_memory=False)
print("Initial shape:", df.shape)

# ===== STEP 2: REMOVE ROWS WHERE URL OR NETBLOCK OWNER IS NULL =====
before = df.shape[0]
df = df.dropna(subset=["URL", "netblock_owner"])
print("Rows removed due to NULL URL or netblock_owner:", before - df.shape[0])

# ===== STEP 3: DEFINE COLUMN GROUPS =====
lexical_cols = [
    "url_length", "num_dots", "has_hyphen", "has_at_symbol",
    "path_length", "query_length", "num_params", "tld_length",
    "uses_https", "url_depth", "has_ip", "domain_length",
    "num_subdomains", "contains_suspicious_word", "hostname_length",
    "has_port", "num_special_chars", "is_shortened",
    "fragment_length", "double_slash_pos",
    "digit_to_length_ratio", "num_unique_chars",
    "has_hex_chars", "file_extension_length",
    "num_underscores", "class"
]

network_cols = [
    "netblock_owner",
    "asn",
    "reverse_dns_present",
    "organisation",
    "dnsssec"
]

# ===== STEP 4: REMOVE LEXICAL-FAILURE ROWS =====
before = df.shape[0]
df = df.dropna(subset=lexical_cols, how="all")
print("Removed lexical-failure rows:", before - df.shape[0])

# ===== STEP 5: REMOVE NETWORK-FAILURE ROWS =====
before = df.shape[0]
df = df.dropna(subset=network_cols, how="all")
print("Removed network-failure rows:", before - df.shape[0])

# ===== STEP 6: REMOVE DUPLICATE URLs =====
before = df.shape[0]
df = df.drop_duplicates(subset=["URL"], keep="first")
print("Duplicate URLs removed:", before - df.shape[0])

# ===== STEP 7: WRITE CLEAN CSV (QUOTE EVERYTHING) =====
df.to_csv(OUTPUT_CSV, index=False, quoting=csv.QUOTE_ALL)
print("\n✅ Clean dataset written to:", OUTPUT_CSV)

# ===== STEP 8: SANITY CHECK =====
print("\nFinal shape:", df.shape)
print("\nRemaining NULL counts (top 10 columns):")
print(df.isna().sum().sort_values(ascending=False).head(10))
```
